# MPI untuk pengelolaan data parallel

Dalam pengelolaan data parallel sistem terbagi menjadi manager core dan worker core dimana, para worker core yang akan mengerjakan tugas yang nantinya akan di kumpulkan ke manager core.

Kunci utama dalam parallel adalah peningkatan kecepatan



## Amdahl’s Law

Amdahl’s Law adalah hukum dalam komputasi paralel yang digunakan untuk memperkirakan percepatan (speedup) maksimal suatu program jika sebagian dari program tersebut dijalankan secara paralel menggunakan banyak prosesor.

Konsep Dasar
Sebuah program biasanya terdiri dari:
1. Bagian serial (tidak bisa diparalelkan) → harus dijalankan secara berurutan.

2. Bagian paralel (bisa dibagi ke banyak prosesor) → dapat dieksekusi bersamaan.

Karena ada bagian serial, maka percepatan kinerja tidak akan pernah linier sempurna meskipun jumlah prosesor sangat banyak.

Rumus
$$ S(N) = \frac{1}{(1-P)+\frac{P}{N}}$$

Dengan
- S(N) = Speedup (percepatan) dengan N prosesor
- P = fraksi program yang bisa di paralelkan (0<P<1)
- N = jumlah prosesor
- (1-P) = bagian serial program

Inti dari Amdahl’s Law

- Bagian serial dari program menjadi bottleneck (pembatas utama percepatan).
- Menambah jumlah prosesor tidak selalu memberikan peningkatan performa signifikan.
- Untuk mendapatkan percepatan lebih besar, yang harus diminimalkan adalah bagian serial.

>Bottleneck artinya bagian yang paling lambat atau paling terbatas dalam suatu proses, sehingga membatasi kinerja keseluruhan. Istilah ini diambil dari leher botol: meskipun badan botol lebar, aliran cairan tetap lambat karena sempitnya leher botol.

## Gustafson

**Hukum Gustafson** menjelaskan **hubungan antara jumlah prosesor dan percepatan (speedup)** dalam sistem **komputasi paralel**, **dengan asumsi ukuran masalah juga bertambah**.

**Inti konsep:**

> *“Dengan menambah jumlah prosesor, kita dapat memperbesar ukuran masalah yang dapat diselesaikan dalam waktu yang sama.”*
> — John L. Gustafson (1988)

Jadi, Gustafson menolak pandangan Amdahl yang mengatakan *speedup terbatas*, karena dalam dunia nyata, **kita biasanya memperbesar ukuran masalah seiring bertambahnya core** — bukan menjaga ukuran masalah tetap.

**Rumus**

$$ S(p) = p - \alpha (p-1) $$

di mana:

| Simbol     | Arti                                                                      |
| ---------- | ------------------------------------------------------------------------- |
| ( S(p) )   | Speedup pada ( p ) prosesor                                               |
| ( p )      | Jumlah prosesor                                                           |
| ( \alpha ) | Fraksi waktu eksekusi yang **tidak dapat diparalelkan (sequential part)** |


* Jika **sebagian kecil kode bersifat sequential** (tidak bisa paralel), sisanya ((1 - \alpha)) bisa berjalan **paralel penuh** di (p) prosesor.
* Gustafson berargumen: jika kita menambah ukuran masalah, maka bagian paralel akan **mendominasi waktu eksekusi**, membuat speedup hampir linier terhadap (p).


Contoh Perhitungan

Misalkan:

* Jumlah prosesor ( p = 8 )
* Fraksi sequential ( \alpha = 0.05 ) (5% tidak bisa diparalelkan)

Maka:
[
S(8) = 8 - 0.05(8 - 1) = 8 - 0.35 = 7.65
]

Artinya, dengan 8 prosesor, speedup ≈ **7.65 kali lebih cepat** dari eksekusi serial.


Perbandingan dengan Hukum Amdahl

| Aspek          | **Hukum Amdahl**                                 | **Hukum Gustafson**                                |
| -------------- | ------------------------------------------------ | -------------------------------------------------- |
| **Fokus**      | Fixed problem size (ukuran masalah tetap)        | Scalable problem size (masalah ikut membesar)      |
| **Perspektif** | Meningkatkan kecepatan program yang sama         | Menyelesaikan masalah lebih besar dalam waktu sama |
| **Rumus**      | ( S(p) = \frac{1}{\alpha + \frac{1-\alpha}{p}} ) | ( S(p) = p - \alpha(p - 1) )                       |
| **Kesimpulan** | Ada batas speedup maksimal                       | Speedup bisa mendekati jumlah prosesor             |
| **Filosofi**   | Pesimis terhadap paralelisme                     | Optimis terhadap skalabilitas                      |

 Analogi:

* **Amdahl**: “Kita ingin menyelesaikan tugas lama lebih cepat.”
* **Gustafson**: “Kita ingin menyelesaikan tugas *lebih besar* dalam waktu yang sama.”

**Hukum Gustafson sangat relevan untuk komputasi HPC, GPU, dan paralel modern**, seperti:

* **PyCUDA / OpenCL / CUDA C** → karena data dan domain simulasi membesar (misal simulasi neutron, CFD, ML).
* **MPI / cluster computing** → di mana penambahan node dipakai untuk memperbesar skala masalah, bukan sekadar mempercepat masalah kecil.

1. HPC Cluster (High Performance Computing Cluster)

Tujuan: memberikan kekuatan komputasi sangat besar dengan menggabungkan banyak komputer (node) menjadi satu sistem.

Digunakan untuk perhitungan ilmiah, simulasi fisika/nuklir, AI/ML skala besar, cuaca/iklim, dll.

Komponennya:
- Head node / login node → tempat user submit job.
- Compute nodes → mesin tempat job dijalankan (CPU/GPU).
- Job scheduler → seperti SLURM / PBS / TORQUE untuk mengatur antrian job.
- Interconnect cepat → misalnya Infiniband, supaya komunikasi antar-node cepat.

Karakteristik:
- Fokus pada kecepatan (throughput).
- Biasanya tidak terlalu peduli ketersediaan 100% uptime (kalau 1 node mati, job bisa di-reshcedule).

Analogi: seperti sekumpulan pekerja laboratorium yang bareng-bareng mengerjakan eksperimen berat secara paralel.


2. HA Cluster (High Availability Cluster)

Tujuan: menjaga ketersediaan layanan (service uptime) agar tidak down.

Biasanya dipakai untuk server aplikasi, database, layanan web, sistem keuangan, dll.

Komponennya:

- Beberapa server (node) yang menjalankan aplikasi sama.
- Load balancer untuk distribusi request.
- Failover system → jika 1 node/server rusak, node lain langsung ambil alih.

Karakteristik:
- Fokus pada keandalan & uptime (availability 24/7).
- Biasanya bukan untuk komputasi berat, tapi untuk menjaga agar layanan selalu tersedia.

Analogi: seperti rumah sakit dengan banyak dokter jaga. Kalau satu dokter sakit, yang lain langsung ganti supaya layanan pasien tetap jalan.

1. GPU (Graphics Processing Unit)

Awalnya dibuat untuk grafis, tapi sekarang dipakai untuk komputasi paralel (CUDA, OpenCL, ROCm, dsb).

Punya ribuan core kecil yang bisa menjalankan operasi sederhana secara masif.

Kelebihan GPU
- Sangat bagus untuk komputasi paralel masif (misalnya matrix operations, deep learning, simulasi Monte Carlo).
- Throughput tinggi → bisa memproses banyak data sekaligus.
- Lebih hemat energi per operasi (FLOP/Watt lebih efisien).
- Cocok untuk workload dengan pola SIMD (Single Instruction, Multiple Data).

Kekurangan GPU
- Tidak optimal untuk tugas serial atau yang butuh logika kompleks per thread.
- Punya memori terbatas (VRAM lebih kecil dibanding RAM server CPU).
- Overhead transfer data CPU ↔ GPU (bisa jadi bottleneck).
- Perlu pemrograman khusus (CUDA, OpenCL) yang lebih rumit dibanding pemrograman CPU biasa.

2. CPU (Central Processing Unit)

General-purpose, dengan sedikit core (biasanya 4–64 di server modern), tapi tiap core sangat kuat dan fleksibel.

Cocok untuk komputasi serial dan beragam jenis instruksi.

Kelebihan CPU
- Lebih baik untuk tugas serial dan kompleks.
- Fleksibel → bisa jalankan OS, aplikasi, logika bercabang.
- Memori lebih besar dan akses data lebih fleksibel.
- Tidak butuh coding khusus untuk paralel sederhana (OpenMP, multiprocessing Python, dsb).

Kekurangan CPU
- Jumlah core terbatas, jadi tidak efisien untuk paralelisasi masif.
- Throughput lebih rendah dibanding GPU dalam komputasi numerik intensif.
- Performa FLOP/Watt lebih rendah daripada GPU untuk workload paralel.

GPU unggul kalau pekerjaanmu: komputasi numerik, AI, matrix operations, simulasi paralel masif.

CPU unggul kalau pekerjaanmu: kontrol alur program, logika bercabang, komputasi serial, memory-heavy tasks.

> Memory-heavy task adalah jenis pekerjaan komputasi yang lebih banyak dibatasi oleh kapasitas dan kecepatan memori (RAM/VRAM) dibanding oleh kekuatan komputasi (CPU/GPU FLOPs). Artinya, bottleneck utamanya bukan di “seberapa cepat prosesor menghitung”, tapi di seberapa cepat data bisa diambil/dikirim dari/ke memori.

- Rank & Communicator: setiap proses punya rank di dalam sebuah communicator (mis. MPI.COMM_WORLD). Pesan hanya “cocok” jika communicator, tag, dan source sesuai.
- Tag & Status: tag membantu membedakan jenis pesan; Status memberi info source, tag, dan jumlah byte diterima.
- Matching: satu Recv mencocokkan tepat satu Send.
- Deadlock umum: dua proses saling Send (blocking) tanpa Recv yang siap; atasi dengan urutan yang konsisten atau pakai non-blocking.

# Tipe komunikasi MPI

## Point-to-Point (dua pihak)
Komunikasi langsung satu pengirim ↔ satu penerima.

````python
from mpi4py import MPI
comm = MPI.COMM_WORLD; rank = comm.Get_rank()

if rank == 0:
    data = {'x': 42}
    req = comm.Isend(data, dest=1, tag=0)
    req.Wait()  # pastikan selesai
else:
    data = None
    req = comm.Irecv(source=0, tag=0)
    data = req.wait()
````

## Collective communication

`MPI_Broadcast`
- Fungsi: Mengirim satu data dari root ke semua proses.
- Hasil: Semua proses memiliki data yang sama.
- Analogi: Dosen mengumumkan nilai ke seluruh mahasiswa.

`MPI_Scatter`
- Fungsi: Membagi data besar di root jadi potongan kecil ke tiap proses.
- Hasil: Tiap proses mendapat bagian data berbeda.
- Analogi: Dosen membagikan soal ujian berbeda ke setiap mahasiswa.

`MPI_Gather`
- Fungsi: Mengumpulkan data mentah dari semua proses ke satu proses root.
- Hasil: Proses root punya array yang berisi semua data dari semua proses. Tidak ada operasi matematika yang dilakukan.
- Analoginya: Semua orang kirim laporan ke sekretaris; sekretaris cuma menumpuk semua laporan apa adanya.

`MPI_Reduce`
- Fungsi: Menggabungkan data dari semua proses dengan operasi tertentu (misalnya penjumlahan, maksimum, minimum, perkalian) ke satu proses root.
- Hasil: Proses root punya satu nilai hasil operasi gabungan.
- Analoginya: Semua orang kirim angka; sekretaris langsung menjumlahkannya, hasil akhir disimpan di sekretaris, bukan daftar angka per orang

````python
# file: demo_mpi.py
from mpi4py import MPI

comm = MPI.COMM_WORLD
rank = comm.Get_rank()   # id proses
size = comm.Get_size()   # jumlah proses

# setiap proses punya nilai sendiri
value = rank + 1   # misalnya proses 0 punya 1, proses 1 punya 2, dst.

# ---- GATHER ----
all_values = comm.gather(value, root=0)
if rank == 0:
    print("Hasil GATHER di root:", all_values)
    # root dapat list: [1, 2, 3, ...] sesuai jumlah proses

# ---- REDUCE ----
total = comm.reduce(value, op=MPI.SUM, root=0)
if rank == 0:
    print("Hasil REDUCE (SUM) di root:", total)
    # root dapat jumlah total semua value

# ---- ALLREDUCE (opsional) ----
total_all = comm.allreduce(value, op=MPI.SUM)
print(f"Proses {rank}: total_all = {total_all}")
# semua proses langsung dapat hasil jumlah total
````

Operator bawaan MPI
- `MPI.SUM` : Penjumlahan
- `MPI.PROD` : Perkalian
- `MPI.MAX` :	Nilai maksimum
- `MPI.MIN` :	Nilai minimum
- `MPI.LAND` : Logical AND
- `MPI.LOR` :	Logical OR
- `MPI.LXOR` : Logical XOR
- `MPI.BAND` : Bitwise AND
- `MPI.BOR` : Bitwise OR
- `MPI.BXOR` : Bitwise XOR
- `MPI.MAXLOC` : Nilai maksimum beserta lokasi indeksnya
- `MPI.MINLOC` : Nilai minimum beserta lokasi indeksnya

Selain operator bawaan (SUM, MAX, MIN, dll.), kita bisa membuat operator kustom kalau kita butuh cara khusus untuk menggabungkan data antar proses.

Konsep Operator Kustom MPI
- MPI bekerja dengan reduce: setiap proses punya data → data itu digabung berpasangan dengan operator → hasilnya dibawa lagi → sampai ke root.
- Operator bawaan hanya mendukung operasi standar.
- Kalau kita punya tipe data khusus (misalnya struct) atau aturan gabungan unik (misalnya rata-rata, penggabungan string), kita bisa definisikan sendiri.

````python
from mpi4py import MPI
import numpy as np

comm = MPI.COMM_WORLD
rank = comm.Get_rank()

# Contoh: operator rata-rata (custom)
def average_op(invec, inoutvec, datatype):
    """
    invec: buffer input dari proses lain
    inoutvec: buffer akumulasi (diubah in-place)
    datatype: tipe MPI
    """
    inoutvec[:] = (inoutvec + invec) / 2.0  # rata-rata dua-duanya

# Buat operatornya
avg_op = MPI.Op.Create(average_op, commute=True)

# data tiap proses
data = np.array([rank+1.0])  # misalnya proses 0=1.0, 1=2.0, dst.

# lakukan reduce dengan operator custom
result = np.array([0.0])
comm.Reduce(data, result, op=avg_op, root=0)

if rank == 0:
    print("Hasil rata-rata custom:", result)
````

# Arsitektur Komputasi Parallel

Arsitektur komputasi paralel adalah desain sistem di mana beberapa prosesor bekerja secara bersamaan untuk menyelesaikan tugas dan masalah yang rumit. Dengan memecah tugas besar menjadi sub-tugas yang lebih kecil dan mengerjakannya pada waktu yang sama, arsitektur ini dapat meningkatkan kecepatan dan efisiensi komputasi.



Ada beberapa arsitektur utama yang digunakan dalam komputasi paralel:
1. **Klasifikasi Flynn**
Klasifikasi ini mengelompokkan arsitektur berdasarkan aliran instruksi (instruksi tunggal atau banyak) dan aliran data (data tunggal atau banyak).
- **SISD ( Single Instruction, Single Data)**: Arsitektur tradisional yang menggunakan satu prosesor untuk mengeksekusi satu aliran instruksi pada satu aliran data pada satu waktu (komputasi serial). Contoh : CPU tunggal
- **SIMD ( Single Instruction, Multiple Data)**: Menggunakan satu aliran instruksi untuk beroperasi pada banyak aliran data secara bersamaan. Arsitektur ini cocok untuk tugas yang membutuhkan operasi identik pada kumpulan data yang besar, seperti pemrosesan gambar atau simulasi ilmiah. Contoh : GPU, prosesor vektor
- **MISD ( Multiple Instruction, Single Data)**: Beberapa prosesor mengeksekusi instruksi berbeda pada aliran data yang sama. Arsitektur ini jarang digunakan dan tidak umum dijumpai. Contoh : Sistem avionik
- **MIMD ( Multiple Instruction, Multiple Data)**: Beberapa prosesor mengeksekusi instruksi berbeda pada aliran data yang berbeda secara bersamaan. Ini adalah arsitektur yang paling umum untuk komputasi paralel modern dan mencakup sistem multiprosesor dan multikomputer. Contoh : Multiprosesor, multicore modern

2. Berdasarkan pembagian memori
Arsitektur ini berfokus pada bagaimana prosesor-prosesor yang berbeda mengakses memori.
- **Arsitektur memori bersama (Shared Memory)**

$\;\;\;\;\;\;$ Cara kerja: Beberapa prosesor berbagi akses ke satu ruang memori tunggal.

$\;\;\;\;\;\;$ Komunikasi: Prosesor berkomunikasi dengan membaca dan menulis dari kumpulan memori yang sama. Komunikasi relatif mudah dan cepat.

$\;\;\;\;\;\;$ Kelemahan: Mengalami kesulitan untuk melakukan penskalaan ( scaling ) ke jumlah prosesor yang sangat besar karena adanya masalah bandwidth dan tabrakan memori.

$\;\;\;\;\;\;$ Contoh: Komputer dengan banyak inti ( multi-core ) yang berbagi memori utama yang sama.

- **Arsitektur memori terdistribusi (Distributed Memory)**

$\;\;\;\;\;\;$ Cara kerja: Setiap prosesor memiliki memori lokalnya sendiri yang tidak dapat diakses langsung oleh prosesor lain.

$\;\;\;\;\;\;$ Komunikasi: Prosesor berkomunikasi dengan mengirimkan pesan ( message passing ) melalui jaringan interkoneksi.

$\;\;\;\;\;\;$ Kelebihan: Sangat mudah untuk melakukan penskalaan karena memori tidak digunakan bersama, sehingga tidak ada masalah bandwidth atau tabrakan.

$\;\;\;\;\;\;$ Kelemahan: Pemrogram harus secara eksplisit mendistribusikan data dan mengelola protokol komunikasi.

$\;\;\;\;\;\;$ Contoh: Klaster komputasi dengan beberapa komputer yang terhubung melalui jaringan.

- **Arsitektur hibrida (Hybrid)**

$\;\;\;\;\;\;$ Cara kerja: Menggabungkan elemen arsitektur memori bersama dan memori terdistribusi.

$\;\;\;\;\;\;$ Pendekatan: Biasanya, setiap node dalam sistem memiliki beberapa prosesor dengan memori bersama, sedangkan komunikasi antar node dilakukan menggunakan message passing.

$\;\;\;\;\;\;$ Keuntungan: Menggabungkan keunggulan paralelisme memori bersama di dalam setiap node dengan skalabilitas memori terdistribusi di beberapa node.

$\;\;\;\;\;\;$ Contoh: Sistem klaster yang setiap komputernya memiliki prosesor multi-inti.

CPU didominasi oleh tiga arsitektur utama: x86, ARM, dan RISC-V

1. x86
Arsitektur x86 dirancang dengan filosofi Complex Instruction Set Computer (CISC).
Ini berarti satu instruksi (perintah) dalam arsitektur x86 dapat melakukan beberapa operasi kompleks sekaligus, seperti mengambil data dari memori, melakukan perhitungan, dan menyimpan hasilnya kembali ke memori.  

Pabrikan utama:
- Intel: Merupakan pengembang asli arsitektur x86. Saat ini, mereka menawarkan prosesor seri Core (misalnya, Core i5, i7, i9, dan Core Ultra) untuk PC, serta seri Xeon untuk server dan stasiun kerja.
- AMD: Awalnya mengadopsi arsitektur x86, dan kini bersaing ketat dengan Intel. Jajaran produknya meliputi seri Ryzen untuk PC (misalnya, Ryzen 5, 7, 9) dan seri Threadripper serta Epyc untuk server dan stasiun kerja.

Karakteristik arsitektur:
- Set Instruksi yang Kompleks: CISC memungkinkan satu instruksi untuk melakukan beberapa operasi, sehingga efisien dalam hal kepadatan kode.
- Kompatibilitas Mundur: Prosesor yang lebih baru dapat menjalankan kode yang ditulis untuk prosesor x86 yang lebih lama.
- Performa Tinggi: Terus-menerus disempurnakan untuk performa komputasi berat, gaming, dan pemrosesan paralel tingkat tinggi.

Cara kerja:
Set Instruksi Kompleks: Prosesor x86 memiliki set instruksi yang sangat besar dan beragam. Instruksinya memiliki panjang yang bervariasi, memungkinkan satu instruksi untuk melakukan pekerjaan yang lebih rumit dibandingkan dengan arsitektur lain.
- Mikrokode: Untuk mengeksekusi instruksi yang kompleks, prosesor x86 modern menggunakan lapisan internal yang disebut mikrokode. Ketika sebuah instruksi x86 diterima, mikrokode akan memecahnya menjadi beberapa operasi yang lebih kecil dan sederhana, mirip dengan instruksi RISC, sebelum dieksekusi.
- Kompatibilitas Mundur: Salah satu fitur utama x86 adalah kompatibilitas mundurnya yang luar biasa. Ini berarti prosesor x86 modern dapat menjalankan perangkat lunak yang ditulis untuk prosesor x86 yang jauh lebih tua, memastikan ekosistem perangkat lunak yang sangat stabil dan luas.

Kelebihan dan kekurangan:
- Kelebihan: Dominasi pasar PC dan server, ekosistem perangkat lunak yang luas, dan performa tinggi untuk tugas komputasi berat.
- Kekurangan: Konsumsi daya yang lebih tinggi dan desain yang lebih kompleks daripada arsitektur RISC.

2. ARM
Arsitektur ARM (sebelumnya Acorn RISC Machine, kini Advanced RISC Machine) didasarkan pada filosofi Reduced Instruction Set Computer (RISC). Filosofi ini menekankan kesederhanaan, menggunakan set instruksi yang lebih kecil dan sederhana untuk mengeksekusi satu operasi dasar per siklus.

Pabrikan utama:
- Apple Silicon: Apple mendesain System on a Chip (SoC) berbasis ARM (misalnya, seri M1, M2, dan M3) untuk Mac, iPad, dan iPhone mereka. Desain ini menawarkan integrasi yang ketat antara perangkat keras dan lunak, sehingga memberikan performa dan efisiensi daya yang luar biasa.
- Qualcomm: Produsen utama chip berbasis ARM untuk perangkat Android, dengan produk andalannya yaitu seri Snapdragon.
- MediaTek: Pesaing utama Qualcomm di pasar perangkat seluler, yang juga memproduksi SoC berbasis ARM.
Arm Holdings: Perusahaan yang mendesain arsitektur ARM dan memberikan lisensi kepada produsen lain. Mereka menawarkan berbagai inti prosesor, seperti seri Cortex-A untuk performa tinggi, Cortex-M untuk daya sangat rendah, dan Neoverse untuk server.

Karakteristik arsitektur:
- Set Instruksi Sederhana: RISC menggunakan set instruksi yang lebih kecil dan sederhana, sehingga memproses instruksi lebih cepat per siklus.
- Efisiensi Daya: Desain ini membutuhkan daya lebih rendah, sehingga ideal untuk perangkat bertenaga baterai.
Modularitas: Arsitektur ini memungkinkan kustomisasi untuk berbagai aplikasi, mulai dari perangkat wearable hingga pusat data.

Cara kerja:
- Set Instruksi Sederhana: Prosesor ARM memiliki set instruksi yang seragam dan lebih sedikit. Meskipun setiap instruksi hanya melakukan satu tugas sederhana, eksekusinya sangat cepat.
- Efisiensi Daya: Karena instruksinya sederhana, prosesor ARM membutuhkan daya lebih sedikit untuk beroperasi. Ini menjadikannya pilihan ideal untuk perangkat yang bergantung pada baterai, seperti smartphone dan tablet.
- Desain Pipeline: Prosesor ARM menggunakan teknik pipelining untuk meningkatkan kecepatan dengan mengeksekusi beberapa instruksi yang berbeda pada saat yang bersamaan, meski setiap instruksi tetap sederhana.
- Optimasi untuk Perangkat Seluler: Produsen chip seperti Apple dan Qualcomm mengoptimalkan desain ARM dengan menambahkan inti pemrosesan khusus untuk tugas-tugas seperti kecerdasan buatan (AI) dan pemrosesan grafis, yang sangat meningkatkan performa dan efisiensi.

Kelebihan dan kekurangan:
- Kelebihan: Efisiensi energi yang superior, biaya produksi rendah, dan performa per watt yang tinggi.
- Kekurangan: Ekosistem perangkat lunak desktop yang lebih kecil (memerlukan emulasi untuk aplikasi x86).

3. RISC-V
RISC-V adalah arsitektur Reduced Instruction Set Computer (RISC) yang bersifat standar terbuka dan bebas royalti. Artinya, siapa pun dapat mengimplementasikan, memodifikasi, dan menggunakan arsitektur ini tanpa perlu membayar lisensi.

Pabrikan utama:
- RISC-V International: Organisasi nirlaba yang mengawasi standar dan pengembangan arsitektur.
- SiFive: Perusahaan yang memproduksi inti prosesor RISC-V dan alat pengembangan.
- NVIDIA: Menggunakan RISC-V untuk kontroler internal di dalam GPU mereka.

Karakteristik arsitektur:
- Open Source: Tidak ada biaya lisensi, sehingga mendorong inovasi dan kolaborasi.
- Modularitas dan Ekstensibilitas: Memungkinkan perancang untuk menyesuaikan set instruksi sesuai dengan kebutuhan aplikasi tertentu.
- Adaptabilitas: Cocok untuk berbagai aplikasi, mulai dari sistem embedded hingga superkomputer.

Cara kerja:
- Modular dan Fleksibel: Arsitektur RISC-V memiliki set instruksi inti yang minimal dan wajib. Pengembang dapat menambahkan ekstensi opsional sesuai kebutuhan, seperti ekstensi untuk operasi perkalian/pembagian (M), operasi atomik (A), atau instruksi vektor (V).
- Dapat Disesuaikan: Desainer memiliki fleksibilitas penuh untuk menyesuaikan prosesor RISC-V untuk aplikasi tertentu, mengoptimalkan daya, performa, dan biaya, mulai dari sistem embedded hingga superkomputer.
- Transparansi dan Kolaborasi: Sifat sumber terbukanya mendorong inovasi dan kolaborasi dari seluruh komunitas. Ini berbeda dari model berpemilik yang ada di x86 dan ARM.

Kelebihan dan kekurangan:
- Kelebihan: Tidak ada biaya lisensi, fleksibilitas tinggi, dan kustomisasi yang luas.
- ekurangan: Ekosistem perangkat lunak dan toolchain masih dalam tahap perkembangan, belum sekomprehensif x86 dan ARM.

Organisasi memori pada arsitektur **MIMD (Multiple Instruction, Multiple Data)** tidak bersifat tunggal; ia terbagi menjadi dua kategori utama yang sangat berbeda: **Shared Memory** (Memori Bersama) dan **Distributed Memory** (Memori Terdistribusi).

MIMD sendiri berarti setiap prosesor dalam sistem dapat menjalankan instruksi yang berbeda pada data yang berbeda secara bersamaan. Cara mereka mengakses memori untuk mendapatkan data inilah yang mendefinisikan organisasinya.

Berikut penjelasan dari kedua model tersebut.

## 1. Shared Memory (Memori Bersama)

Dalam model ini, semua prosesor (atau inti) terhubung ke satu ruang alamat memori global yang sama.

* **Konsep Utama:** Setiap prosesor dapat membaca dan menulis ke lokasi memori manapun secara langsung.
* **Analogi:** Bayangkan sebuah tim yang bekerja di depan satu **papan tulis raksasa (whiteboard)** yang sama. Siapa pun bisa menulis atau membaca dari bagian manapun di papan tulis itu.
* **Komunikasi:** Komunikasi antar prosesor terjadi secara *implisit* melalui memori. Jika Prosesor A ingin mengirim data ke Prosesor B, Prosesor A hanya perlu menulis data tersebut ke lokasi memori yang telah disepakati, dan Prosesor B kemudian membacanya.
* **Tantangan:** Tantangan utamanya adalah **sinkronisasi** dan **cache coherence**. Karena semua orang bisa menulis di papan tulis yang sama, harus ada aturan (seperti *locks* atau *semaphores*) untuk mencegah dua prosesor menulis di tempat yang sama pada waktu yang bersamaan (disebut *race condition*).

Model Shared Memory ini dibagi lagi menjadi dua jenis:

### a. UMA (Uniform Memory Access)
* **Artinya:** Waktu akses (latensi) dari prosesor manapun ke lokasi memori manapun adalah **sama** (seragam).
* **Contoh:** Sebagian besar PC dan laptop *multi-core* modern. Inti CPU 1 membutuhkan waktu yang sama untuk mengakses RAM seperti halnya Inti CPU 2.
* **Skalabilitas:** Baik untuk jumlah prosesor yang sedikit, tetapi menjadi *bottleneck* (macet) jika prosesornya terlalu banyak karena semua berebut jalur yang sama ke memori.

### b. NUMA (Non-Uniform Memory Access)
* **Artinya:** Waktu akses **tidak sama**. Sistem masih memiliki satu ruang alamat global, tetapi memori secara fisik terbagi-bagi dan ditempatkan dekat dengan prosesor tertentu.
* **Cara Kerja:** Prosesor dapat mengakses memori "lokal" (yang terpasang padanya) dengan sangat cepat, tetapi akan jauh lebih lambat saat mengakses memori "remote" (yang terpasang pada prosesor lain).
* **Contoh:** Server *high-end* dan superkomputer. Ini adalah cara untuk membangun sistem *shared memory* dengan jumlah prosesor yang jauh lebih banyak (skalabilitas lebih baik daripada UMA).


## 2. Distributed Memory (Memori Terdistribusi)

Dalam model ini, setiap prosesor memiliki memori privatnya sendiri yang **tidak dapat diakses** secara langsung oleh prosesor lain.

* **Konsep Utama:** Tidak ada ruang alamat memori global. Setiap prosesor hanya "melihat" memorinya sendiri.
* **Analogi:** Bayangkan sekumpulan orang yang bekerja di **kantor pribadi** masing-masing. Setiap orang punya meja dan catatan (memori) sendiri.
* **Komunikasi:** Komunikasi terjadi secara *eksplisit* melalui **Message Passing (Pengiriman Pesan)**. Jika Prosesor A ingin mengirim data ke Prosesor B, ia harus mengemas data itu ke dalam "pesan" dan secara aktif mengirimkannya melalui jaringan interkoneksi. Prosesor B kemudian harus secara aktif "menerima" pesan tersebut.
* **Contoh:** *Cluster* komputer (di mana banyak komputer terpisah dihubungkan oleh jaringan), atau superkomputer masif (MPP - Massively Parallel Processors).
* **Skalabilitas:** Sangat tinggi. Anda bisa terus menambahkan "kantor" (prosesor + memori) baru ke dalam jaringan.
* **Tantangan:** Lebih sulit diprogram karena programmer harus mengelola semua pengiriman dan penerimaan data secara manual (misalnya menggunakan *library* seperti **MPI - Message Passing Interface**).



## 3. Hybrid (Gabungan)

Pada praktiknya, arsitektur paling umum saat ini adalah **hibrida**:

* Sistem ini terdiri dari banyak *node* (komputer individual).
* **Di dalam satu node:** Menggunakan arsitektur **Shared Memory** (biasanya NUMA). Beberapa prosesor/inti berbagi memori lokal di node tersebut.
* **Di antara node-node:** Menggunakan arsitektur **Distributed Memory**. Setiap node berkomunikasi dengan node lain menggunakan *message passing* melalui jaringan.

Superkomputer modern hampir semuanya adalah *cluster* hibrida. Programmer biasanya menggunakan model hibrida pula: **MPI** untuk komunikasi antar-node, dan **OpenMP** (atau *threading*) untuk paralelisasi di dalam node (memanfaatkan *shared memory*).

## Akses memori pada shared memori

Berikut adalah penjelasan untuk setiap arsitektur akses memori tersebut:

## UMA (Uniform Memory Access)
* **Apa itu:** Model arsitektur di mana semua prosesor (inti) memiliki **waktu akses yang seragam (sama)** ke lokasi manapun di memori utama (RAM).
* **Analogi:** Sebuah meja bundar besar (memori) di mana semua orang (prosesor) yang duduk mengelilinginya dapat mengambil benda apapun di atas meja dengan waktu dan usaha yang sama.
* **Cara Kerja:** Terdapat satu *memory bus* atau *controller* utama yang digunakan bersama oleh semua prosesor untuk mengakses satu *pool* memori utama.
* **Konteks "Default":** Komentar Anda benar. Sebagian besar komputer pribadi (*desktop* dan *laptop*) menggunakan model ini. CPU *multi-core* Anda (misalnya, 4 inti) semuanya terhubung ke *controller* memori yang sama untuk mengakses RAM.
* **Kelebihan:** Model pemrograman sederhana (memori bersama).
* **Kekurangan:** Tidak dapat diskalakan (semakin banyak prosesor, semakin padat *bus* memori, menyebabkan kemacetan).


## NUMA (Non-Uniform Memory Access)
* **Apa itu:** Model arsitektur di mana sistem masih memiliki satu ruang alamat memori global (semua prosesor bisa melihat semua memori), namun waktu aksesnya **tidak seragam**.
* **Analogi:** Satu perusahaan besar (sistem) dengan beberapa gedung (node). Setiap gedung memiliki ruang arsip (memori) sendiri. Karyawan (prosesor) di Gedung A bisa mengambil arsip di gedungnya dengan **sangat cepat (local access)**. Jika dia butuh arsip dari Gedung B, dia bisa mengambilnya (shared address space), tapi butuh waktu **lebih lama (remote access)** karena harus berjalan antar gedung.
* **Cara Kerja:** Memori secara fisik didistribusikan ke dalam "node". Setiap node memiliki prosesor dan memori lokalnya sendiri. Akses ke memori lokal cepat; akses ke memori di node lain lebih lambat karena harus melalui **interconnect**.
* **Penggunaan:** Umum digunakan pada *server high-end* dan superkomputer untuk mengatasi *bottleneck* UMA.


## NORMA (No Remote Memory Access)
* **Apa itu:** Model arsitektur **memori terdistribusi (distributed memory)** murni. Di sini, **tidak ada ruang alamat memori global**. Setiap prosesor hanya dapat mengakses memorinya sendiri (memori lokal).
* **Analogi:** Sekelompok rumah terpisah (prosesor). Setiap rumah memiliki kotak surat dan perpustakaan pribadi (memori lokal). Anda **tidak bisa** masuk ke rumah tetangga untuk mengambil buku. Jika Anda butuh buku dari tetangga, Anda harus mengirim surat (pesan) meminta buku itu, dan tetangga Anda harus secara aktif mengambil buku itu dan mengirimkannya kembali kepada Anda.
* **Cara Kerja:** Komunikasi antar prosesor hanya bisa terjadi secara eksplisit melalui **message passing (pengiriman pesan)**. Satu prosesor harus mengirim (Send) data, dan prosesor lain harus menerima (Receive) data tersebut.
* **Penggunaan:** Digunakan dalam *cluster* komputer masif di mana banyak komputer independen dihubungkan oleh jaringan (misalnya, *Beowulf clusters*) dan menjadi dasar dari model pemrograman **MPI (Message Passing Interface)**.

## COMA (Cache Only Memory Architecture)
* **Apa itu:** Model yang lebih kompleks (dan jarang digunakan) yang mencoba menggabungkan kemudahan *shared memory* (seperti UMA/NUMA) dengan skalabilitas *distributed memory* (seperti NORMA).
* **Analogi:** Bayangkan sebuah tim di mana tidak ada lemari arsip pusat (main memory). Sebaliknya, setiap anggota tim (prosesor) memiliki meja yang sangat besar (cache) yang menampung semua data yang mereka butuhkan.
* **Cara Kerja:** Tidak ada "memori utama" yang statis. Seluruh memori sistem diperlakukan sebagai *cache*. Data tidak memiliki "rumah" permanen; ia **bermigrasi** dan **direplikasi** secara otomatis ke *cache* (memori lokal) dari prosesor yang sedang menggunakannya. Perangkat keras khusus (disebut *directory*) melacak di mana salinan data berada untuk menjaga konsistensi.
* **Tujuan:** Menyederhanakan pemrograman (terlihat seperti *shared memory*), tetapi secara dinamis memindahkan data agar selalu lokal (seperti *distributed memory*), untuk mendapatkan performa tinggi.
* **Kekurangan:** Sangat rumit untuk diimplementasikan di perangkat keras, sehingga tidak umum digunakan secara komersial.

Arsitektur heterogen, atau *heterogeneous computing*, mengacu pada sistem komputer yang menggunakan lebih dari satu jenis prosesor atau inti pemrosesan (core). Alih-alih mengandalkan satu jenis Unit Pemrosesan Pusat (CPU) untuk semua tugas, sistem ini menggabungkan berbagai jenis prosesor yang masing-masing memiliki keunggulan untuk tipe pekerjaan tertentu.

Tujuan utamanya adalah untuk mencapai **kinerja yang lebih tinggi** dan **efisiensi daya yang lebih baik** dengan mendelegasikan tugas kepada prosesor yang paling cocok untuk menanganinya.

## Komponen Utama

Sistem dengan arsitektur heterogen biasanya terdiri dari kombinasi beberapa prosesor berikut:

- **CPU (Central Processing Unit):** Prosesor serbaguna yang handal dalam menangani tugas-tugas sekuensial (berurutan) dan pekerjaan umum sistem operasi. CPU ibarat seorang manajer proyek yang menangani berbagai tugas administratif dan mengoordinasikan pekerjaan.
- **GPU (Graphics Processing Unit):** Awalnya dirancang untuk grafis, GPU memiliki ribuan inti kecil yang sangat efisien untuk tugas-tugas paralel (banyak pekerjaan serupa yang dijalankan bersamaan). Ini membuatnya ideal untuk kecerdasan buatan (AI), pembelajaran mesin (*machine learning*), dan analisis data besar. GPU seperti tim pekerja yang melakukan ribuan tugas repetitif secara serentak.
- **DSP (Digital Signal Processor):** Sangat efisien dalam memproses sinyal digital seperti audio, video, atau data dari sensor. DSP unggul dalam operasi matematika yang repetitif dan kompleks.
- **FPGA (Field-Programmable Gate Array):** Sirkuit terintegrasi yang dapat diprogram ulang oleh pengguna setelah diproduksi. Ini memberikan fleksibilitas tinggi dan akselerasi untuk tugas-tugas yang sangat spesifik.
- **ASIC (Application-Specific Integrated Circuit):** Cip yang dirancang untuk melakukan satu tugas spesifik dengan sangat cepat dan efisien. Contohnya adalah cip untuk *mining* mata uang kripto.

## Cara Kerja dan Analogi

Bayangkan sebuah dapur restoran. Daripada memiliki satu koki serbaguna yang melakukan segalanya mulai dari memotong sayuran, memanggang roti, hingga memasak steak, dapur heterogen memiliki beberapa spesialis:
* Seorang koki utama (**CPU**) yang mengawasi operasi dan menangani pesanan kompleks.
* Beberapa asisten koki (**GPU**) yang secara bersamaan memotong sayuran dalam jumlah besar.
* Seorang pembuat roti (**DSP/ASIC**) yang hanya fokus membuat adonan dan memanggang dengan sangat efisien.

Dengan membagi pekerjaan seperti ini, seluruh dapur menjadi jauh lebih cepat dan efisien. Sistem komputasi heterogen bekerja dengan prinsip yang sama: sistem operasi atau perangkat lunak akan secara cerdas mengalokasikan setiap tugas ke prosesor yang paling efisien untuk menyelesaikannya.

## Keuntungan dan Tantangan

### Keuntungan
* **Peningkatan Kinerja:** Dengan menggunakan prosesor yang tepat untuk tugas yang tepat, kecepatan pemrosesan secara keseluruhan meningkat drastis.
* **Efisiensi Daya:** Tugas dapat diselesaikan dengan konsumsi energi yang lebih rendah karena dijalankan pada perangkat keras yang dioptimalkan untuk tugas tersebut. Ini sangat penting untuk perangkat seluler dan laptop.
* **Fleksibilitas:** Memungkinkan perancang sistem untuk menyeimbangkan antara biaya, kinerja, dan konsumsi daya sesuai kebutuhan aplikasi.

### Tantangan
* **Kompleksitas Pemrograman:** Mengembangkan perangkat lunak yang dapat secara efektif memanfaatkan berbagai jenis prosesor lebih sulit daripada memprogram untuk satu jenis CPU saja.
* **Manajemen Memori:** Memastikan semua prosesor dapat mengakses data yang mereka butuhkan secara efisien tanpa konflik bisa menjadi tantangan teknis.
* **Overhead Komunikasi:** Diperlukan waktu dan sumber daya untuk memindahkan data dan instruksi antar prosesor yang berbeda, yang dapat mengurangi sebagian keuntungan kinerja.


## Contoh di Kehidupan Sehari-hari

Arsitektur heterogen sudah ada di sekitar kita:

* **Ponsel Pintar (Smartphone):** Ponsel modern adalah contoh utama. Di dalamnya terdapat CPU untuk tugas umum, GPU untuk game dan antarmuka, DSP untuk pemrosesan audio dan sinyal seluler, serta prosesor khusus untuk kamera dan AI.
* **Konsol Game:** Perangkat seperti PlayStation dan Xbox menggunakan kombinasi CPU dan GPU yang kuat untuk menghasilkan grafis game yang realistis dan gameplay yang lancar.
* **Superkomputer:** Pusat data dan superkomputer modern sering kali menggabungkan ribuan CPU dan GPU untuk menangani simulasi ilmiah, riset AI, dan analisis data skala besar.
* **Laptop Modern:** Banyak laptop kini dilengkapi dengan arsitektur "big.LITTLE" pada CPU-nya, yang menggunakan inti berperforma tinggi (*performance cores*) untuk tugas berat dan inti berefisiensi tinggi (*efficiency cores*) untuk tugas ringan guna menghemat baterai.

Menggunakan MPI (Message Passing Interface) pada arsitektur heterogen memiliki alur yang sedikit berbeda dari arsitektur homogen (yang semua prosesornya sama). Fokus utamanya bergeser dari "membagi pekerjaan secara merata" menjadi "**membagi pekerjaan secara adil dan cerdas**" sesuai kemampuan setiap prosesor.

Pada dasarnya, **MPI tetap bertindak sebagai 'kurir' atau 'layanan pos'** yang mengirimkan pesan dan data antar proses. MPI tidak peduli apakah proses pengirim adalah CPU dan proses penerima adalah GPU; tugasnya hanya mengantarkan pesan dari Peringkat (Rank) X ke Peringkat Y.

Tantangan sebenarnya terletak pada *bagaimana Anda memprogram* alur tersebut.


## Alur Paralelisasi MPI pada Arsitektur Heterogen

Berikut adalah alur kerja umumnya, seringkali menggunakan pola *Master-Worker*:

### 1. Inisialisasi dan Identifikasi (Tahap Kritis)
1.  **`MPI_Init`:** Semua proses (di CPU, GPU, dll.) memulai lingkungan MPI.
2.  **`MPI_Comm_size` & `MPI_Comm_rank`:** Setiap proses mendapatkan jumlah total proses (`size`) dan ID uniknya (`rank`).
3.  **Identifikasi Perangkat Keras (Pembeda Utama):** Ini adalah langkah yang *tidak* ada di sistem homogen. Setiap proses (rank) harus mencari tahu "Saya ini siapa?".
    * Sebuah proses bisa mengecek: "Apakah saya memiliki akses ke perangkat CUDA/NVIDIA?" Jika ya, ia mengidentifikasi dirinya sebagai *worker* GPU.
    * Jika tidak, ia mengidentifikasi dirinya sebagai *worker* CPU.
4.  **Komunikasi Kemampuan:** Setiap *worker* (Peringkat 1, 2, 3, ...) mengirimkan pesan `MPI_Send` ke proses *Master* (biasanya Peringkat 0) yang berisi identitasnya (misal: "Saya CPU, kecepatan saya X" atau "Saya GPU, kecepatan saya Y"). Proses Master menggunakan `MPI_Recv` untuk mengumpulkan informasi ini.

### 2. Partisi Tugas & Load Balancing (Oleh Master)
1.  **Pembagian Tidak Merata:** Proses Master (Peringkat 0) kini tahu persis siapa *worker*-nya dan seberapa cepat mereka. Ia *tidak* akan membagi pekerjaan secara merata.
2.  **Analogi:** Bayangkan Master punya 1000 surat untuk dilipat.
    * Ia tahu Peringkat 1 (CPU) bisa melipat 100 surat per jam.
    * Ia tahu Peringkat 2 (GPU) bisa melipat 900 surat per jam.
3.  **Distribusi Cerdas:** Master *tidak* akan mengirim 500 ke Peringkat 1 dan 500 ke Peringkat 2. Ia akan mengirim 100 surat ke Peringkat 1 (CPU) dan 900 surat ke Peringkat 2 (GPU).
4.  **Alat MPI:** Alih-alih `MPI_Scatter` (yang mengirim data dalam potongan yang sama), Master akan menggunakan **`MPI_Scatterv`** (v = vector/variable), yang memungkinkannya mengirimkan potongan data dengan **ukuran yang berbeda-beda** ke setiap proses.



### 3. Eksekusi Paralel (Di Masing-Masing Perangkat)
1.  Setiap proses *worker* menerima bagian pekerjaannya (misal, via `MPI_Scatterv`).
2.  **Peringkat 1 (CPU):** Menjalankan fungsi C++/Fortran standar untuk memproses 100 suratnya.
3.  **Peringkat 2 (GPU):**
    * Menyalin 900 suratnya dari memori CPU ke memori GPU.
    * Meluncurkan *kernel* CUDA atau OpenCL untuk memproses 900 surat itu secara paralel di ribuan inti GPU.
    * Menyalin hasilnya kembali dari memori GPU ke memori CPU.
4.  Selama fase ini, MPI mungkin tidak banyak digunakan, kecuali para *worker* perlu bertukar data di tengah jalan (misalnya, `MPI_Send` / `MPI_Recv` antar *worker*).

### 4. Pengumpulan Hasil (Gathering)
1.  Setelah selesai, setiap *worker* (Peringkat 1 dan 2) mengirimkan hasilnya kembali ke Master (Peringkat 0).
2.  **Alat MPI:** Master akan menggunakan **`MPI_Gatherv`** (v = vector/variable) untuk mengumpulkan hasil yang ukurannya mungkin berbeda-beda dari setiap *worker*.

### 5. Finalisasi
1.  Master (Peringkat 0) menggabungkan semua hasil parsial (hasil 100 surat + hasil 900 surat) menjadi satu solusi akhir.
2.  **`MPI_Finalize`:** Semua proses mengakhiri lingkungan MPI.


## Tantangan Utama

1.  **Load Balancing:** Ini adalah tantangan terbesar. Menentukan rasio kerja yang tepat antara CPU dan GPU sangat sulit. Jika salah, GPU mungkin selesai terlalu cepat dan 'menganggur' menunggu CPU, atau sebaliknya. Ini menghapus semua keuntungan kinerja.
2.  **Kompleksitas Kode:** Anda harus menulis dua jenis kode dalam satu aplikasi: kode untuk dieksekusi di CPU dan kode (seperti kernel CUDA) untuk dieksekusi di GPU.
3.  **Overhead Transfer Data:** Memindahkan data dari memori utama (CPU) ke memori perangkat (GPU) dan kembali lagi adalah operasi yang sangat "mahal" (lambat). Program harus dirancang untuk **meminimalkan transfer data** dan **memaksimalkan komputasi** pada GPU untuk setiap data yang ditransfer.

## Pola komunikasi pengelolaan data paralel (antar core):
- **Broadcast**, manager core mengirim data yang sama ke semua worker core, tiap worker core akan menerima data yang sama.
- **Scatter**, manager core memecah sebuah set data menjadi beberapa bagian kemudian mengirimkannya ke tiap worker core, jadi tiap worker core akan mendapatkan data yang berbeda beda.
- **Gather**, manager core akan menerima setiap data dari workercore, dan hanya mengumpulkan data dari tiap workercore.
- **Reduce**, managercore menerima data dari workercore dan dapat melakukan operasi logika dan numerik seperti (+-:x).

Ini adalah konsep hierarki fundamental dalam pemrograman GPU, khususnya menggunakan **CUDA (NVIDIA)**, untuk mengelola ribuan (bahkan jutaan) tugas paralel.

Secara singkat, hierarkinya adalah:
* Satu **Grid** adalah kumpulan dari banyak **Block**.
* Satu **Block** adalah kumpulan dari banyak **Thread**.

Bayangkan Anda harus mengelola sebuah **Tentara (Grid)** besar untuk menyelesaikan satu misi.

## 1. Thread (Prajurit)
* **Apa itu:** Ini adalah unit eksekusi terkecil. Anggap saja sebagai **satu prajurit**.
* **Tugasnya:** Setiap *thread* menjalankan instruksi atau fungsi yang Anda tulis (disebut *kernel*).
* **Contoh:** Jika tugas Anda adalah mencerahkan setiap piksel pada gambar, satu *thread* bertanggung jawab untuk menghitung nilai satu piksel saja.

## 2. Block (Peleton)
* **Apa itu:** Ini adalah sekelompok *thread*. Anggap saja sebagai **satu peleton** yang terdiri dari banyak prajurit.
* **Fitur Utama:** Thread di dalam *block* yang sama dapat **bekerja sama** satu sama lain.
* **Cara Bekerja Sama:**
    1.  **Shared Memory:** Mereka memiliki akses ke "papan tulis" kecil yang sangat cepat (memori bersama) yang hanya bisa dilihat oleh anggota peleton itu.
    2.  **Sinkronisasi:** Anda bisa memerintahkan semua *thread* di *block* itu untuk "berhenti dan tunggu" di titik tertentu sampai semua rekan mereka selesai (`__syncthreads()`).
* **Contoh:** Satu *block* (misalnya 256 *thread*) mungkin ditugaskan untuk memproses satu bagian kecil (misalnya, 16x16 piksel) dari gambar. Mereka bekerja sama untuk memuat data ke *shared memory* agar lebih efisien.

## 3. Grid (Tentara)
* **Apa itu:** Ini adalah kumpulan dari semua *block*. Anggap saja sebagai **seluruh tentara** yang terdiri dari banyak peleton.
* **Tugasnya:** Satu kali Anda menjalankan fungsi/program di GPU (*kernel launch*), Anda meluncurkan satu *grid*.
* **Fitur Utama:** Semua *block* di dalam *grid* menjalankan instruksi yang sama, tetapi pada data yang berbeda. Namun, *block* satu **tidak bisa berkomunikasi** atau berkoordinasi secara langsung dengan *block* lain. Mereka bekerja secara independen.
* **Contoh:** Seluruh *grid* (misalnya, 4000 *block*) ditugaskan untuk memproses keseluruhan gambar 1920x1080 piksel.

### Ringkasan

| Unit | Analogi | Deskripsi | Bisa Bekerja Sama? |
| :--- | :--- | :--- | :--- |
| **Thread** | Prajurit | Unit eksekusi terkecil. | Tidak (secara individu). |
| **Block** | Peleton | Sekelompok *thread*. | **Ya**, *thread* di dalam *block* bisa berbagi memori & sinkronisasi. |
| **Grid** | Tentara | Sekelompok *block*. | **Tidak**, *block* di dalam *grid* bekerja secara independen. |

Manajemen memori di antara ketiganya (Threads, Blocks, Grid) bersifat **hierarkis**. Setiap level memiliki akses ke jenis memori yang berbeda, dengan kecepatan dan ruang lingkup (visibilitas) yang juga berbeda.

Ini adalah salah satu konsep terpenting dalam pemrograman GPU karena kinerja program Anda sangat bergantung pada seberapa efisien Anda mengelola data di berbagai memori ini.

Berikut adalah rinciannya, dari yang terkecil hingga terbesar:


## 1. Memori untuk `Thread` (Prajurit)

Setiap *thread* memiliki memori "pribadi"-nya sendiri yang tidak dapat dilihat oleh *thread* lain.

* **Registers (Register):**
    * **Apa itu:** Memori terkecil dan **tercepat** di GPU.
    * **Cara Pakai:** Digunakan secara otomatis oleh *compiler* untuk menyimpan variabel lokal di dalam fungsi Anda (misal: `int i = 0;`). Programmer tidak mengelolanya secara langsung.
    * **Visibilitas:** Hanya bisa dilihat oleh *thread* itu sendiri.

* **Local Memory (Memori Lokal):**
    * **Apa itu:** Ini adalah "cadangan" untuk *thread*. Jika sebuah *thread* kehabisan Register (misalnya, Anda membuat *array* lokal yang besar), data itu akan tumpah (*spill*) ke Memori Lokal.
    * **Peringatan:** Namanya menipu. Memori Lokal secara fisik berlokasi di **Global Memory** (VRAM), yang **sangat lambat**. Sebisa mungkin, hindari penggunaan Memori Lokal.
    * **Visibilitas:** Hanya bisa dilihat oleh *thread* itu sendiri.


## 2. Memori untuk `Block` (Peleton)

Semua *thread* di dalam *block* yang sama berbagi satu jenis memori khusus.

* **Shared Memory (Memori Bersama):**
    * **Apa itu:** Memori *on-chip* yang **sangat cepat** (hampir secepat Register).
    * **Cara Pakai:** Dideklarasikan secara eksplisit oleh programmer (misal: `__shared__ int cache[256];`).
    * **Tujuan Utama:** Ini adalah "papan tulis" untuk peleton. *Thread* dalam satu *block* menggunakannya untuk berkolaborasi, bertukar data, dan menyimpan data yang sering diakses dari Global Memory (berfungsi sebagai *cache* yang dikelola manual).
    * **Visibilitas:** Semua *thread* di dalam *block* yang sama bisa melihatnya. *Block* lain tidak bisa melihatnya.
    * **Sinkronisasi:** Penggunaannya **wajib** disertai dengan sinkronisasi (`__syncthreads()`) untuk memastikan semua *thread* melihat data yang konsisten.


## 3. Memori untuk `Grid` (Tentara)

Ini adalah memori yang dapat diakses oleh **semua** *thread* dari **semua** *block* yang ada di dalam *grid*.

* **Global Memory (Memori Global):**
    * **Apa itu:** Ini adalah memori utama GPU (VRAM). Ini adalah memori yang **terbesar** namun juga yang **terlambat**.
    * **Cara Pakai:** Ini adalah tempat Anda menyalin data dari CPU (Host) sebelum *kernel* dieksekusi, dan tempat Anda menyimpan hasil akhir untuk disalin kembali ke CPU.
    * **Visibilitas:** Dapat diakses (baca/tulis) oleh semua *thread* di *grid* dan juga oleh CPU (Host).

* **Constant Memory (Memori Konstan):**
    * **Apa itu:** Memori khusus yang dioptimalkan untuk data yang **hanya dibaca** (*read-only*) oleh GPU.
    * **Cara Pakai:** Dideklarasikan di level *grid* dan datanya tidak dapat diubah oleh GPU.
    * **Visibilitas:** Semua *thread* di *grid* bisa membacanya.
    * **Keuntungan:** Sangat cepat (karena di-*cache*) jika semua *thread* membaca data yang sama (misal: konstanta fisika, koefisien filter).

* **Texture Memory (Memori Tekstur):**
    * **Apa itu:** Mirip dengan Constant Memory (di-*cache*, *read-only*), tetapi dioptimalkan secara khusus untuk pola akses spasial (misalnya, mengambil data piksel yang berdekatan dalam pemrosesan gambar).

## Ringkasan Manajemen Memori

| Jenis Memori | Lingkup / Visibilitas | Kecepatan | Dipakai Untuk |
| :--- | :--- | :--- | :--- |
| **Registers** | 1 `Thread` | **Sangat Cepat** | Variabel lokal (otomatis) |
| **Local Memory** | 1 `Thread` | **Sangat Lambat** | Tumpahan register (dihindari) |
| **Shared Memory** | 1 `Block` | **Sangat Cepat** | Kolaborasi antar *thread*, *cache* manual |
| **Global Memory** | 1 `Grid` (dan CPU) | **Lambat** | Input data utama dan output hasil |
| **Constant Memory** | 1 `Grid` (dan CPU) | **Cepat (di-cache)** | Data *read-only* (misal: konstanta) |

## Multiprocessing

 Komponen Utama

* **CPU (Core)**: Setiap proses dapat berjalan di core yang berbeda.
* **Process**: Setiap proses memiliki **memori sendiri (isolated memory space)**.
* **IPC (Inter-Process Communication)**: Mekanisme untuk bertukar data antar proses seperti `Pipe`, `Queue`, atau `SharedMemory`.

Cara Kerja

Multiprocessing membuat **beberapa proses independen** yang berjalan secara **paralel** di CPU berbeda.
Setiap proses memiliki:

* interpreter Python sendiri,
* ruang memori sendiri (tidak berbagi data langsung).

Contoh sederhana (Python):

```python
from multiprocessing import Process

def worker(n):
    print(f"Proses {n} sedang berjalan")

if __name__ == '__main__':
    for i in range(4):
        p = Process(target=worker, args=(i,))
        p.start()
```

 Komunikasi

* Menggunakan **IPC (Inter-Process Communication)**: `Queue`, `Pipe`, `Manager`, atau file bersama.
* Komunikasi relatif lambat karena melibatkan **serialisasi data** (pickling).

Kapan Digunakan

* Saat tugas bersifat **CPU-bound**, seperti:

  * komputasi numerik besar (misal simulasi reaktor, integral numerik, dll),
  * image processing berat,
  * perhitungan paralel dengan CPU banyak.

 Kelebihan

* Tidak terpengaruh oleh GIL (Global Interpreter Lock).
* Bisa memanfaatkan banyak core CPU sepenuhnya.
* Lebih stabil untuk tugas berat (isolasi proses).

 Kekurangan

* Konsumsi memori besar (tiap proses punya salinan sendiri).
* Komunikasi antar proses lebih lambat.
* Overhead tinggi untuk membuat banyak proses.


##  Multithreading

Komponen Utama

* **Thread**: Jalur eksekusi kecil dalam satu proses.
* **Shared Memory**: Semua thread berbagi ruang memori proses utama.
* **GIL (Global Interpreter Lock)** di Python membatasi thread agar hanya satu thread Python aktif di satu waktu.

Cara Kerja

Multithreading membuat **beberapa thread** dalam **satu proses**.
Thread berbagi memori yang sama dan dapat bekerja secara **asinkron**.

Contoh sederhana (Python):

```python
import threading

def worker(n):
    print(f"Thread {n} berjalan")

threads = []
for i in range(4):
    t = threading.Thread(target=worker, args=(i,))
    t.start()
```

Komunikasi

* Sangat cepat, karena semua thread **berbagi memori**.
* Namun, perlu **sinkronisasi** dengan `Lock`, `Semaphore`, atau `Event` agar tidak terjadi **race condition**.

Kapan Digunakan

* Untuk tugas **I/O-bound**, seperti:

  * membaca/menulis file,
  * menunggu data jaringan (web scraping, socket),
  * operasi input/output yang banyak menunggu.

Kelebihan

* Lebih hemat memori dibanding multiprocessing.
* Komunikasi antar thread cepat.
* Cocok untuk I/O-bound task.

Kekurangan

* Tidak bisa memanfaatkan multi-core secara penuh (karena GIL di Python).
* Risiko deadlock dan race condition.
* Debugging lebih sulit.


## MPI (Message Passing Interface)

Komponen Utama

* **Node (komputer/mesin)**: Tiap node bisa memiliki banyak core.
* **Process**: Tiap node menjalankan proses MPI.
* **Communicator (COMM_WORLD)**: Mengatur semua proses yang berkomunikasi.

Cara Kerja

MPI digunakan untuk **komputasi terdistribusi** di beberapa mesin (cluster).
Setiap proses berjalan secara **independen** dan **berkomunikasi eksplisit** lewat **message passing**.

Contoh (Python via `mpi4py`):

```python
from mpi4py import MPI
comm = MPI.COMM_WORLD
rank = comm.Get_rank()
size = comm.Get_size()

print(f"Halo dari proses {rank} dari total {size}")
```

Komunikasi

* Proses saling kirim data lewat **message passing**: `send()`, `recv()`, `bcast()`, `scatter()`, `gather()`, `reduce()`.
* Tidak berbagi memori sama sekali (setiap proses terpisah).

Kapan Digunakan

* Saat menjalankan program di **cluster komputer** atau **HPC (High Performance Computing)**.
* Ideal untuk **simulasi ilmiah besar**, **reaktor nuklir**, **AI training paralel**, atau **Monte Carlo parallel**.

Kelebihan

* Dapat dijalankan di banyak node (bukan hanya 1 CPU).
* Skalabilitas tinggi untuk superkomputer.
* Tidak bergantung pada GIL.

Kekurangan

* Pengaturan dan debugging kompleks.
* Harus didefinisikan komunikasi eksplisit antar proses.
* Tidak efisien untuk data kecil atau komunikasi sering.

 Perbandingan Langsung

| Aspek                     | **Multithreading**                | **Multiprocessing**            | **MPI**                        |
| ------------------------- | --------------------------------- | ------------------------------ | ------------------------------ |
| **Tipe paralelisme**      | Thread-level (dalam 1 proses)     | Process-level (antar core CPU) | Distributed-level (antar node) |
| **Memori**                | Shared (berbagi memori)           | Isolated (terpisah)            | Isolated (antar mesin)         |
| **Komunikasi**            | Cepat (langsung)                  | Lewat IPC                      | Lewat pesan MPI (send/recv)    |
| **Overhead**              | Rendah                            | Sedang                         | Tinggi                         |
| **Kinerja CPU-bound**     | Rendah (GIL)                      | Tinggi                         | Sangat tinggi                  |
| **Kinerja I/O-bound**     | Tinggi                            | Rendah-sedang                  | Tidak cocok                    |
| **Lingkup eksekusi**      | 1 mesin                           | 1 mesin multi-core             | Multi-node cluster             |
| **Kompleksitas**          | Mudah                             | Sedang                         | Sulit                          |
| **Contoh library Python** | `threading`, `concurrent.futures` | `multiprocessing`, `pathos`    | `mpi4py`                       |

### \#\# Arsitektur Jaringan di Linux

Arsitektur jaringan di Linux mengikuti model berlapis (layered model) yang mirip dengan model TCP/IP. Ini memungkinkan pemisahan fungsi, sehingga setiap lapisan hanya perlu berinteraksi dengan lapisan di atas dan di bawahnya.

Secara sederhana, alurnya seperti ini dari atas ke bawah saat mengirim data:

1.  **Aplikasi Pengguna (User-space):** Ini adalah program yang Anda gunakan, seperti web browser (Firefox, Chrome) atau terminal. Aplikasi ini ingin mengirim atau menerima data.
2.  **Socket Layer:** Ini adalah jembatan antara aplikasi dan kernel (inti sistem operasi). Ketika aplikasi ingin terhubung ke jaringan, ia membuat "socket".
3.  **Network Stack (Kernel-space):** Ini adalah inti dari pemrosesan jaringan di Linux.
      * **Protocol Stacks (TCP/UDP):** Data dari socket dibungkus sesuai protokol transport yang dipilih (misalnya, TCP untuk koneksi yang andal, UDP untuk koneksi yang cepat).
      * **IP Layer:** Paket data dari TCP/UDP kemudian dibungkus lagi dengan header IP, yang berisi alamat IP sumber dan tujuan. Di sinilah *routing* (penentuan rute) terjadi.
      * **Link Layer (Device Drivers):** Akhirnya, paket IP diserahkan ke driver kartu jaringan (Network Interface Card/NIC). Driver ini yang berkomunikasi langsung dengan perangkat keras untuk mengirimkan data sebagai sinyal listrik atau gelombang radio.
4.  **Perangkat Keras (Hardware):** Kartu jaringan (NIC) fisik yang mengirimkan data melalui kabel Ethernet atau Wi-Fi.

Saat menerima data, prosesnya terjadi secara terbalik.

### \#\# DNS (Domain Name System)

**DNS adalah** sistem yang berfungsi seperti buku telepon untuk internet. Tugas utamanya adalah menerjemahkan nama domain yang mudah diingat manusia (contoh: `www.google.com`) menjadi alamat IP yang dimengerti oleh komputer (contoh: `142.250.191.132`).

  * **Analogi:** Anda lebih mudah mengingat nama "Toko Buku Gramedia" daripada mengingat alamat lengkapnya di "Jalan Merdeka No. 43". DNS melakukan hal yang sama untuk website.
  * **Cara Cek di Linux:** Gunakan perintah `dig` atau `nslookup`.
    ```bash
    dig google.com
    ```

### \#\# Switch

**Switch adalah** perangkat keras jaringan yang menghubungkan beberapa perangkat (komputer, printer, server) dalam satu **jaringan lokal (LAN)**.

  * **Cara Kerja:** Switch bekerja pada Layer 2 (Data Link Layer). Ia "pintar" karena dapat mengenali alamat fisik (MAC Address) dari setiap perangkat yang terhubung padanya. Saat data masuk, switch hanya akan meneruskannya ke port tujuan yang spesifik, bukan ke semua port.
  * **Analogi:** Resepsionis di sebuah gedung kantor yang tahu persis di meja mana setiap karyawan duduk. Ketika ada kiriman untuk "Budi", resepsionis akan langsung mengantarkannya ke meja Budi, bukan mengumumkannya ke seluruh kantor.


### \#\# Router

**Router adalah** perangkat keras jaringan yang menghubungkan **dua atau lebih jaringan yang berbeda**. Fungsi utamanya adalah meneruskan (me-*routing*) paket data dari satu jaringan ke jaringan lainnya, misalnya dari jaringan lokal Anda ke jaringan internet.

  * **Cara Kerja:** Router bekerja pada Layer 3 (Network Layer). Ia menggunakan alamat IP untuk menentukan jalur terbaik bagi paket data untuk mencapai tujuannya.
  * **Analogi:** Pengatur lalu lintas di bandara. Ia mengarahkan pesawat (paket data) dari satu landasan (jaringan lokal) ke rute penerbangan yang benar untuk mencapai bandara tujuan (jaringan lain).


### \#\# Broadcast, Gateway, dan Loopback

  * **Broadcast :** Pesan yang dikirim dari satu perangkat ke **semua perangkat lain** yang berada dalam satu segmen jaringan yang sama. Alamat IP broadcast khusus biasanya adalah alamat terakhir dalam sebuah subnet (misalnya, `192.168.1.255`). Ini digunakan saat sebuah perangkat perlu menemukan layanan lain tanpa mengetahui alamat spesifiknya, seperti saat mencari server DHCP.

  * **Gateway :** Pintu gerbang keluar dari jaringan lokal menuju jaringan lain (seperti internet). Biasanya, alamat gateway adalah alamat IP dari router di jaringan Anda. Setiap kali komputer Anda ingin berkomunikasi dengan perangkat di luar jaringannya, ia akan mengirimkan data tersebut ke gateway.

  * **Loopback :** Antarmuka jaringan virtual yang digunakan komputer untuk "berbicara" dengan dirinya sendiri. Alamat IP-nya selalu **127.0.0.1** (disebut juga `localhost`). Ini sangat berguna bagi pengembang untuk menguji aplikasi jaringan pada mesin lokal tanpa perlu terhubung ke jaringan fisik.


### \#\# Subnet

**Subnet (atau Subnetwork) adalah** pembagian logis dari sebuah jaringan IP yang besar menjadi beberapa jaringan yang lebih kecil. Ini dilakukan dengan menggunakan **Subnet Mask**.

  * **Tujuan:**
      * **Efisiensi:** Mengurangi lalu lintas broadcast yang tidak perlu.
      * **Keamanan:** Memisahkan bagian jaringan yang berbeda.
      * **Manajemen:** Memudahkan pengelolaan alamat IP.
  * **Analogi:** Sebuah kota besar (jaringan utama) dibagi menjadi beberapa area dengan kode pos yang berbeda (subnet). Ini membuat pengiriman surat (data) menjadi lebih terorganisir dan efisien. Subnet Mask (`255.255.255.0`) adalah aturan yang menentukan batas-batas kode pos tersebut.

### \#\# Perbedaan IPv4 dan IPv6

IPv6 diciptakan karena alamat IPv4 sudah hampir habis. Berikut adalah perbedaan utamanya:

| Fitur | IPv4 (Internet Protocol version 4) | IPv6 (Internet Protocol version 6) |
| :--- | :--- | :--- |
| **Ukuran Alamat** | 32-bit | 128-bit |
| **Format Alamat** | Desimal bertitik (Contoh: `192.168.1.1`) | Heksadesimal dipisahkan titik dua (Contoh: `2001:0db8:85a3:0000:0000:8a2e:0370:7334`) |
| **Jumlah Alamat** | Sekitar 4,3 miliar | Sangat besar (340 triliun triliun triliun), praktis tidak akan habis |
| **Konfigurasi** | Umumnya manual atau melalui DHCP | Mendukung konfigurasi otomatis (SLAAC) tanpa butuh server DHCP |
| **Keamanan** | IPSec (keamanan) bersifat opsional | IPSec terintegrasi sebagai standar |

#### **Cara Mengecek Alamat IP di Linux**

Anda bisa menggunakan perintah `ip addr` atau `ifconfig` (yang lebih tua).

```bash
ip addr
```

**Contoh Output:**

```

    link/ether 08:00:27:a1:b2:c3 brd ff:ff:ff:ff:ff:ff
    inet 192.168.1.15/24 brd 192.168.1.255 scope global dynamic noprefixroute enp0s3  <-- Ini Alamat IPv4
```

  * `inet`: Menunjukkan alamat IPv4 Anda.
  * `inet6`: Menunjukkan alamat IPv6 Anda.

### \#\# DHCP (Dynamic Host Configuration Protocol)

**DHCP adalah** protokol yang memungkinkan sebuah server untuk secara **otomatis** memberikan alamat IP dan informasi konfigurasi jaringan lainnya (seperti subnet mask, gateway, dan server DNS) kepada perangkat klien dalam sebuah jaringan.

  * **Fungsi:** Menghilangkan kebutuhan untuk mengatur alamat IP setiap perangkat secara manual. Saat Anda menghubungkan laptop ke Wi-Fi baru, DHCP-lah yang membuat Anda langsung bisa terhubung ke internet.
  * **Analogi:** Seorang manajer hotel yang memberikan kunci kamar (alamat IP) kepada setiap tamu (perangkat) yang baru datang. Ketika tamu check-out, kunci kamar tersebut bisa diberikan kepada tamu berikutnya. Ini jauh lebih efisien daripada setiap tamu harus mencari kamar kosong sendiri.
